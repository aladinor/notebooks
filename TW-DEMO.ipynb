# TruWeather Demo

---

Let’s instantiate an Arraylake client, point it at the Earthmover API, and grab the `earthmover-demos/tw-test` repo so we can open it with `xarray`.

```python
from arraylake import Client
import xarray

client = Client()
repo = client.get_repo("earthmover-demos/tw-test")
```

Establish a read-only view of the immutable “main” branch

```python
session = repo.readonly_session("main")
```

Open the repo using `Xarray`

```python
import xarray as xr

dtree = xr.open_datatree(
    session.store,
    zarr_format=3,
    consolidated=False,
    engine="zarr"
)
```

Check the content of this dataset using `dtree.children`

```python
list(dtree.children)
```

## Surface Observations

To access surface observations, we can use a `file-path` syntax. 

```python
dtree["/surf_obs"]
```

```python
list(dtree["surf_obs"].children)
```

Checking **Barani** instrument 

```python
dtree["surf_obs/BARANI"].ds
# dtree["surf_obs/INTEL"].ds
```

We can plot data as follows

```python
# dtree["surf_obs/BARANI"].air_temp.plot(
#     x="time"
# )
dtree["surf_obs/BARANI"].relative_humidity.plot(
    x="time"
)
```

## Radar/Lidar data
Similarly, we can access lidar and radar data

### Denton radar data

Let's check inside Denton dataset

```python
dtree["lidar_radar/denton.tx"]
```

Danton radar data contains only one `sweep`

```python
dtree["lidar_radar/denton.tx/sweep_0"]
```

Plotting radar reflectivity

```python
dtree["lidar_radar/denton.tx/sweep_0"].reflectivity_factor_dbz.plot(
    x="x",
    y="y",
    vmin=-10,
    vmax=60,
    cmap="ChaseSpectral"
)
```

### Lidar data

Check the content of the `L0AD00003JP` lidar instrument

```python
list(dtree["lidar_radar/L0AD00003JP"].children)
```

Now, we can check the data variables within `sweep_0`

```python
dtree["lidar_radar/L0AD00003JP/sweep_0"].data_vars
```

Plotting radial velocity in `sweep_4`

```python
dtree["lidar_radar/L0AD00003JP/sweep_4"].radial_velocity_mps.plot(
    x="x",
    y="y",
    vmin=-10,
    vmax=10,
    cmap="ChaseSpectral"
)
```

Spectrum_v at `sweep_4`

```python
dtree["lidar_radar/L0AD00003JP/sweep_4"].spectrum_v.isel(nfft=60).plot(
    x="x",
    y="y",
    robust=True,
    cmap="ChaseSpectral"
)
```
